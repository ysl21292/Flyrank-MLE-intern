## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

- Question: Which pages should an editor check first because they are losing Google views?
- One row = one page.
- Output: a ranked list of pages. The top page is checked first.
- Who acts and what they do: an editor checks the top pages, then decides to update, protect, or leave each one.
- Cost of a wrong call: wasted editor time on a page that was fine, or a page that keeps losing views and nobody looks at it.
- Why data or ML can help: there are too many pages to check by hand, and one simple rule cannot rank them all.
- Why this is not just "train a model": first I choose the question, the action, and the cost. A model comes later, and it only counts if it beats a simple rule.

```python
visible = df[df.impressions_90d >= 500]
down = visible[visible.trend_direction == "down"]
print(f"Pages with 500+ views (impressions) in 90 days: {len(visible):,}")
print(f"Of those, pages whose views are going down:     {len(down):,}")
print("Too many to check by hand, so the order of the list matters.")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

1. A few pages carry most of the views, so a good "check first" list could matter a lot.
2. More than half the pages are going down, so "going down" alone does not tell me where to start.
3. One simple rule catches almost nothing, so I need a better way to rank. I have not tested that yet.

```python
s = df.impressions_90d.sort_values(ascending=False)
top5 = s.iloc[: int(len(s) * 0.05)].sum() / s.sum()
down_share = (df.trend_direction == "down").mean()
stale = (df.days_since_last_update >= 180) & (df.impressions_90d >= 500)

print(f"1) The top 5% of pages get {top5:.0%} of all views.")
print(f"2) {down_share:.0%} of pages have views going down.")
print(f"3) The simple rule 'not updated in 180+ days' only catches {int(stale.sum())} busy pages.")
```
